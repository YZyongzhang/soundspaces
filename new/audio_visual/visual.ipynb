```python
audio = list()
tag = list()
visual = list()
audio_ = data[0]['audio']
tag_ = data[0]['rl_pred']
visual_ = data[0]['camera']
num = 0
for index,tag__ in enumerate(tag_):
    if tag__ !=3 and tag__ !=0:
        audio.append(audio_[index])
        visual.append(visual_[index])
        tag.append(tag__)
    if tag__ == 0 and num <=4:
        num +=1
        audio.append(audio_[index])
        visual.append(visual_[index])
        tag.append(tag__)
audio.pop(0)
tag.pop(0)
visual.pop(0)
```
