```python
import matplotlib.pyplot as plt
import numpy as np
import pickle
with open("../../data/new/agent_3/offline_episode_3_759.pkl" , 'rb') as f:
    data = pickle.load(f)
img = data[0]['camera']
```

```python
def draw_img(img):
    img = img.astype('uint8')
    plt.imshow(img)
```

```python
img = img[3]
```

```python
draw_img(img)
```

```python
img.shape
```

```python
def Gmask(img):
    height , width , _ = img.shape
    mask = np.ones((height, width),dtype=np.uint8)
    height , width = mask.shape
    for h_i in range(height):
        for w_i in range(width):
            # left
            if h_i + 2* w_i < 128:
                mask[h_i][w_i] = 0
            # right
            if h_i - 2* w_i < -128:
                mask[h_i][w_i] = 0
    return mask
```

```python
mask = Gmask(img)
mask = mask[:, :, np.newaxis]
new_img = mask*img
```

```python
draw_img(new_img)
```
